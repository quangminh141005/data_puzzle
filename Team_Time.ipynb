**I. What is Monte Carlo Simulation:**
- Is a way to model probability of different outcomes that can not be easily predicted due to the intervention of random variable
- Face with significant uncertainty in making a forecast or estimate some methods replace the uncertain variable with a single average number. The Monte Carlo simulation instead uses multiple values and then averages the result
- Today, Monte Carlo simulations are increasingly being used in conjunction with AI models

**II. How Monta Carlo Simulation work**
- The Monte Carlo acknowledges an issue for any simulation technique: The probability of varying outcomes cannot be firmly pinpointed because of random variable interference *(Không được xác định chắc chắn do sự xuất hiện của biến ngẫu nhiên)*. Therefore, a Monte Carlo simulaiton focuses on constantly repeating random samples
- The simulations takes the variable that has uncertainty and assigns it a random value. The model is then run, and the result is provide. This process is repeated again and again while assigning manu different values to the variable in question
- Its like saying:*"I donts knnow the exact answer, but i i try it 100,000,000 times and look at the results, i'll get a very good estimate."*
